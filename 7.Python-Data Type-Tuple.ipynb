```python
# Tuple
# A tuple is similar to a list .The diff is that we can not change the elements of a tuple once it has assigned
 # where we can change the element of list
# In short  ,a tuple ia an immutable list. A tuple can not changrd in any way once it is created

# Charaterstics
 # orderd
 # unchangeable
 # allow duplicate
# Plan of attack
 # Creating Tuple
 # Accessing Item
 # Editing Item
 # Deleting item
 # Operations on tuple
 # Functions of tuple
```

```python
# Creating Tuple
 # empty
T1=()
print(T1)

# Creating tuple with single item
t2=(2,)
print(t2)
print(type(t2))

t3=(1,2,3,(4,5))
print(t3)
print(type(t3))

# type conversion
t4=tuple('hello')
print(t4)
```

```python
# Indexing
t1=(1,2,3)
print(t1)
print(t1[0])
print(t1[-1])

```

```python
# Slicing
t=(1,2,3,4,5)
print(t[1:3])
print(t[0:4])
print(t[0:])
print(t[::-1])
print(t[-3:-1])

t1=(1,2,3,4,(5,6))
print(t1[1:7])
```

```python
# operation + and *

t=(1,2,3,4,5,6)
t1=(7,8,9,10)
print(t+t1)
print(t1*3)

# Membership
print(1 in t)
print(6 in t1)
print( 10 not in t1)

for i in t:
  print(i)
```

```python
# Tuple functions
# len/min/max/sorted/sum

t=(8,7,8,23,25,89,8,95)
print(len(t))
print(sum(t))
print(min(t))
print(max(t))
print(sorted(t))

# count
print(t.count(8))

# Index
print(t.index(95))
```

```python
# Difference between Lisr and tuple

 # Syntax
 # Mutablity
 # Speed
 # Memory
 # Build in funcationality
 # Error prone
 # Usability



import time
L = list(range(100000000))
T = tuple(range(100000000))

start = time.time()
for i in L:
  i*6
print('List Time',time.time()-start)

start = time.time()
for i in T:
  i*6
print('Tuple Time',time.time()-start)
```

```python
# Memory
import sys
L=list(range(100))
T=tuple(range(100))
print('List size',sys.getsizeof(L))
print('Tuple size',sys.getsizeof(T))
```

```python
# Error Prone
a = [1,2,3]
b = a
print(a)
print(b)
a.append(10)
print(a)
print(b)

a = (1,2,3)
b = a
a = a+(4,)
print(a)
print(b)
```

```python
# Tuple unpacking
a,b,c=(1,2,3)
print(a,b,c)

# swaping numbers
a=1
b=2
a,b=b,a
print(a,b)

a,b,*others=(1,2,3,4)
print(a,b)
print(*others)
```
