```python
# Method overloading
# Method overloading not support in pythom
class shape():
  def area(self,r):
    return (3.14*r*r)
  def area(self,l,b):
    return l*b
```

```python
# We can implement method overloading using this way

class shape:
  def area(self,a,b=0):
    if b ==0:
      return 3.14*a*a
    else:
      return  a*b
```

```python
s=shape()
print("Area of circle",s.area(5))
print("Area of rectangle",s.area(5,7))
```
