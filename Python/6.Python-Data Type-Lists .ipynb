```python
# How lists are store in memory

a=2
print(a)
print(id(2))
print(id(a))

L=[1,2,3]
print(id(L)) # total list address

print(id(L[0]))
print(id(L[1]))
print(id(L[2]))

print(id(1))
print(id(2))
print(id(3))
```

```python
# Orderd

L=[1,2,3]
L1=[3,2,1]
print(L==L1)

a=[1,2,3]
b=[1,2,3]
print(a==b)
```

```python
# Hetrogenous
l=[1,1.9,10,True,2+5j]
print(l)
print(id(2+5j))
print(id(l[0]))
print(id(1))
```

```python
# creating List
# empty

L=[]
print(L)

# 1D
L=[1,2,3,4,5,6,8,9]
print(L)

# 2D
L=[1,2,3,[4,5,6]]
print(L)

# hetrogonous
l=[1,True,False,1+2j,-1,4.88]
print(l)

# Type conversion

L=list('hello')
print(L)
print(type(L))
```

```python
# Accessing items from lists
#Indexing
L=[1,2,3,4,5,6]
print(L[0])
print(L[1])
print(L[0:])
print(L[:7])

# print(L[10]) # list index out of range error

# Negative
print(L[-5])
print(L[-1])
print(L[-4])

l=[1,2,3,[4,5]]
print(l)
print(l[3])
print(l[3][0]) # list inside list find 0th
print(l[-1])
print(l[-1][-2])
```

```python
#Slicing

a=[1,2,3,4,5,6]
print(a[:7])
print(a[-5:-1])
print(a[0::2])
print(a[-5:-2:2])
print(a[::-1]) # reverse
```

```python
# Adding element in to the list
s=[1,2,3,3,4,5]
print(s)
print(id(s))
s.append(6)
print(s)
print(id(s))
s.append(True)
print(s)
print(id(s))
s.append([6,7])
print(s)

# Extend
s.extend([6,7])
print(s)

s.extend('pune')
print(s)

s.append('pune')
print(s)

# Insert

e=[1,2,3,4,5]
print(e)
e.insert(1,100)
print(e)
```

```python
# Editing

L=[1,2,3,4,5,6]
print(L)
L[-1]=200
print(L)
L[1:4]=[200,300,400]
print(L)
```

```python
# Deleting items from list
a=[1,2,3,4]
print(a)
del a[1:3]
print(a)

# Remove
l=[1,2,3,4,6]
l.remove(6)
print(l)

# Clear
L=[1,2,3,4,5,6]
print(L)
L.clear()
print(L)

# pop
s=[1,2,3,4,5]
s.pop(3)
print(s)
```

```python
# Operations On list

# Arithmatic operation(+,*)

L1=[1,2,3,4]
L2=[5,6,7,8]

print(L1+L2) #merge
print(L1*3)


# Membership

a=[1,2,3,4,5]
b=[6,7,8]
print(5 in a)
print( 5 not in b)
print( 6 in a)
```

```python
# Loop

L1=[1,2,3,4]
L2=[1,2,3,4,[5,6]]

for i in L2:
  print(i)

for i in L1:
  print(i)

  L2 = [1, 2, 3, 4, [5, 6]]

for i in L2:
    if type(i) == list:
        for j in i:
            print(j)
    else:
        print(i)
```

```python
# List Functions
# len/min/max/sorted

L=[2,1,5,7,8,0]
print(min(L))
print(max(L))
print(len(L))
print(sorted(L))
```

```python
# count()
L=[1,2,3,1,2,4,1,5]
print(L.count(1))
print(L.count(2))
```

```python
# index
L=[1,2,3,1,2,4,1,5]
print(L.index(1))
print(L.index(2))
print(L.index(1,3))
print(L.index(1,6))

# reverse
L.reverse() # permanently reverse the list
print(L)
```

```python
# List Comprehension
# List Comprehension provide a easy way of creating list
# newlist=[expression for time in itrable if condition ==True]

l=[]
for i in range(1,11):
  l.append(i)
print(l)

L=[ i for i in range(1,11)]
print(L)

v=[2,3,4]
s=-3
print(v*s)

# output [-6,-9,-12]
x=[]
for i in v:
  x.append(i*s)
print(x)

x=[s*i for i in v]
print(x)

L=[1,2,3,4,5]
x=[]
for i in L:
  x.append(i*i)
print(x)

x=[i*i for i in L]
print(x)

L=[ i for i in range(1,101) if i%5==0]
print(L)


#Cartesion product
L=[1,2,3,4]
L1=[5,6,7,8]
c=[]

for i in L:
  for j in L1:
    c.append(i*j)
  print(c)

  c=[i*j for i in L for j in L1]
  print(c)

  # Advantage of List comperhension
  +
```
