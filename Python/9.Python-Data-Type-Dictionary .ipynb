```python
# 1D

dict={'name':'shubhangi','age':23,'gender':'Female'}
print(dict)


# 2D

s={
    'name':'shubhangi',
    'College':'MGMs',
    'Sub':{
        'dsa':40,
        'math':80,
        'eng':70
    }}
print(s)
```

```python
# Accessing item

my_dict={
    'name':'shubhangi',
    'College':'MGMs',
    'Sub':{
        'dsa':40,
        'math':80,
        'eng':70
    }}
print(my_dict)
print(my_dict['Sub'])
print(my_dict['name'])


# get
print(my_dict.get('name'))
print(my_dict.get('Sub'))
```

```python
# Adding key value pair
info={'name':'Rajendra','age':50}
print(info)
info['gender']='Male'
print(info)
print(info.get('name'))
```

```python
# Remove key value pair
my_dict={'name':'Rajendra','age':50,'gender':'Male'}
print(my_dict)
my_dict.pop('gender')
print(my_dict)


my_dict={'name':'Rajendra','age':50,'gender':'Male','weight':70}
print(my_dict)
my_dict.popitem()
print(my_dict)

```

```python
# Editing
s={
    'name':'shubhangi',
    'College':'MGMs',
    'Sem':4,
    'Sub':{
        'dsa':40,
        'math':80,
        'eng':70
    }}
print(s)
s['Sem']=5
print(s)
s['Sub']['dsa']=50
print(s)
```

```python
#Dictinory operations

#Membership
s={
    'name':'shubhangi',
    'College':'MGMs',
    'Sem':4,
    'Sub':{
        'dsa':40,
        'math':80,
        'eng':70
    }}
print('shubhangi' in s)
print('name' in s)

for i in s:
  print(i)

for i in s:
  print(i,s[i])
```
