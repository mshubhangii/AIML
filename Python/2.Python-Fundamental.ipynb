4.Operators

* Arthmetic
* Relational
* Logical
* Assignment
* Membership
* Bitwise


```python
#Arthmetic
print(5+9)
print(9-8)
print(9/3)
print(8*9)
print(5//10)
print(5**2)
print(10%5)
```

```python
# Relational
print(5>9)
print(8<9)
print(5>=9)
print(5<=8)
print(6==6)
print(7!=9)
```

```python
#Logical
print(1 and 0)
print(1 or 3)
print(not 0)
```

```python
#Bitwise
print(bin(2 & 3)) # and
print(bin(2 | 3)) # or

print(2^3)
print(~3)
print(4>>2)
print(4<<2)
```

```python
# Assinment
a=3
a+=10
print(a)


a=10
a-=2
print(a)

a=5
a*=10
print(a)

a=10
a/=5
print(a)
```

```python
#Membership
# in / not in
print('p' in 'pune')
print('R' in 'shubhanhi')
print('p'  not in 'Pune')
```
