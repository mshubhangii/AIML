```python
# Menu driven calculator
num1=int(input('enter your num1:'))
num2=int(input('Enter your num2:'))

orp= input('Enter operation:')

if orp =='+':
 print(num1+num2)

elif orp=='-':
 print(num1-num2)

elif orp=='*':
  print(num1*num2)
else:
  print(num1/num2)
```
