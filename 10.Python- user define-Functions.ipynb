```python
# *args and ** kwargs

  # *args and **kwargs are special keywords that are used to pass the variable length argument to a function
```

```python
# * args
# allow us to pass variable number of non keyword argument
```

```python
def add(*args):
    total = 0
    for i in args:
        total = total + i
    return total
```

```python
add(5,9,5)
```

```python
def multiplay(*args):
  product = 1
  for i in args:
    product = product * i
  return product
```

```python
multiplay(10,5,2,3)
```

```python
# How *args work,internaly it create a tuple
# print function also use internally use *args

def multiplay(*args):
  product = 1
  for i in args:
    product = product * i
  print(args)
  print(type(args))
  return product

multiplay(10,5,2,3)
```

```python
# **kwargs
# **kwargs allow us to pass any number of keyword argument
# keyword argument means that they contain a key value pair, like a python dictoinary


def show(**kwargs):
  for (key,value) in kwargs.items():
    print(key,'-->',value)

show(india='Delhi',shrilanka='colombo',pakistan='Karachi')
```

```python
# Points to remember while using *args and **kwargs
 # order of the argument matter(normal ---> *args --> **kwargs)
 # The  works "args" and "kwargs" are only convention , you can use any name as per your choice
```

```python
# Lambda Function
# A lambda function is a small anonymous function
# A lambda function can take any number of argument ,but can only have one expression
```

```python
a=lambda x:x**2
print(a(2))
```

```python
print(a(4))
```

```python
add=lambda x,y:x+y
print(add(4,5))
```

```python
multiplay=lambda x,y:x*y
print(multiplay(4,6))
```

```python
a=lambda x: 'even' if x% 2==0  else 'odd'
print(a(5))
```
