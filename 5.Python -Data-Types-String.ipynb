```python
# Create String

s = 'hello'
print(s)
s = "hello"
print(s)
# Multiline string
s = '''hello'''
print(s)
s = """ hello"""
print(s)
s = str('hello')

print(s)
```

```python
# Accessing String

s = 'hello world'

# Indexing

print(s[0])
print(s[1])
print(s[6])
print(s[5])

# negative indexing
print(s[-1])
print(s[-5])
```

```python
# slicing
# Positive Indexing
s = 'hello world'
print(s[0:5])
print(s[2:3])
print(s[2:])
print(s[:3])
print(s[:])
print(s[0:6:2])
print(s[0:10:3])
print(s[0:])
```

```python
l='Python is a easy language'

print(l[0:])
print(l[0:6:2])
print(l[0:15:2])
print(l[:])
print(l[:25])
print(l[0:10])
print(l[0:7])
print(l[3:4])
print(l[0:18:3])

# Negative
print(l[-5:-1])
print(l[-25:-1])
print(l[-10:-9])
print(l[6:0:-1])
print(l[-10:])
print(l[:-1])
print(l[-8:-1])
```

```python
# Negative Indexing


s = 'hello world'
print(s[6:0:-1])

# Reverse String
print(s[::-1])

print(s[-5:])
```

```python
# Editing & deleting string

s = 'hello world'

print(s[0])
# s[0]='H'
# Python strings are immutable

s = 'hello world'

del s

```

```python
# Operations on string
# Arithmetaic Operation
print('Pune' + 'Mumbai')
print('Pune' +' ' + 'Mumbai')
#print('Pune' - 'Mumbai')
print('Pune'*5)
print('*' * 50)
```

```python
# Relational operator
print('pune' == 'Mumbai')
print('pune' == 'pune')
print('pune' != 'Pune')
print('pune' > 'Pune')
```

```python
# Logical operator

'hello' and 'world'
'hello' or 'world'
```

```python
# String Functions
# len
print(len('hello world'))
# min
print(min('hello world'))
# max
print(max('hello world'))
# sorted
print(sorted('hello world')) # sorted function retuen should be in list
print(sorted('hello world',reverse = True))
```

```python
# Capitalize/Title/Upper/Lower

s = 'Hello World'
print(s.capitalize())
print(s.title())
print(s.upper())
print(s.lower())
print(s.swapcase())
```
