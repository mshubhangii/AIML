```python
# Sets

# A sets is an unordered collection of items.Every set element is unique(no duplicate) and must be mutable
# A set can also used to perform mathmatical operations like union


# Charaterstics
  # Unordered
  # No duplicate
  # Can not contain mutable data type
```

```python
# Creating sets

# empty set

s={}
print(s)
print(type(s))

s1={1,2,3,4}
print(s1)
print(type(s1))

#s2={1,2,3,4,{5,6}}
#print(s2)

#s={1,2,3,4,[5,6]}
#print(s)

s3={1,2,3,4,(5,6)} # hetro
print(s3)

a={1,2,2,4,6,4,9}
print(a)

# hetro

s4={1,'hello',4,5,True,6.7}
print(s4)


# Type conversion

s6=set([1,2,3])
print(s6)

# Duplicate  not allowed

s7={1,1,2,3,3,8,9,10}
print(s7)

# Unorderd

s1={1,2,3}
s2={3,2,1}
print(s1==s2)
```

```python
# Adding items

s={1,2,3}
print(s)
s.add(50)
print(s)

s1={1,2,3,4}
print(s1)
s1.update([9,8,7])
print(s1)
```

```python
# Deleting
s={1,2,3,4}
print(s)
#del
print(s)

# Discard
s.discard(3)
print(s)

s.discard(7)
print(s)

# Remove
x={1,2,3,4,5}
print(x)
print(x.remove(5))
print(x)

#x.remove(8) # it will throw error

# pop

y1={1,2,3,4,5,6,7}
y1.pop()
print(y1)
```

```python
# Clear
z={1,2,3,4,5}
print(z)
z.clear()
print(z)
```

```python
#  Set operation

s1={1,2,3,4}
s2={3,4,5,6}

# union
print(s1|s2)

#intersection
print(s1&s2)

# Difference
print(s1-s2)
print(s2-s1)
```

```python
# Set functions
# len/sum/mim/max/sorted

s={3,4,5,2,7,8}
print(len(s))
print(sum(s))
print(min(s))
print(max(s))
print(sorted(s))
```

```python
# issubset/issuperset/isdisjoint

s1={1,2,3,4}
s2={3,4,5,5}

print(s1.isdisjoint(s2)) # if all items are differenr in set then in ser then it will return True otherwise it will return false

s3={1,2,3,4}
s4={7,8,9,10}
print(s3.isdisjoint(s4))

# issubset
s5={1,2,3,4,5,6}
s6={3,4,5}

print(s5.issubset(s6))
print

# issuperset
```

```python
# Frozenset

fs=frozenset([1,2,3])
print(fs)

s=frozenset((1,2,3))
print(s)

s1=frozenset({1,2,3})
print(s1)
```
